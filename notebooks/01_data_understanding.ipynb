```python
profile = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.values,
    "Missing Values": df.isnull().sum().values,
    "Missing %": (df.isnull().sum() / len(df) * 100).round(2).values,
    "Unique Values": df.nunique().values
})

profile
```

```python
profile.to_csv("../data/processed/data_profile.csv", index=False)
```

```python
profile.sort_values(by="Missing %", ascending=False).head(20)
```

```python
import matplotlib.pyplot as plt

target_counts = df["TARGET"].value_counts()

plt.figure(figsize=(6,4))
plt.bar(target_counts.index.astype(str), target_counts.values)

plt.title("Loan Default Distribution")
plt.xlabel("TARGET")
plt.ylabel("Number of Customers")

plt.show()
```

```python
df["CODE_GENDER"].value_counts()
```

```python
gender_counts = df["CODE_GENDER"].value_counts()

plt.figure(figsize=(6,4))
plt.bar(gender_counts.index, gender_counts.values)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")

plt.show()
```

```python
df["NAME_EDUCATION_TYPE"].value_counts()
```

```python
education = df["NAME_EDUCATION_TYPE"].value_counts()

plt.figure(figsize=(8,5))
plt.bar(education.index, education.values)

plt.xticks(rotation=45)

plt.title("Education Level")

plt.show()
```
